```python
   import requests

   with open("../conf/api_key.txt", encoding="utf-8") as f:
       API_KEY = f.read().strip()
   with open("../conf/url.txt", encoding="utf-8") as f:
       url = f.read().strip()

   headers = {"Authorization": f"Bearer {API_KEY}"}
   payload = {
       "model": "gpt-oss-120b",
       "messages": [{"role": "user", "content": "Say hello in one short sentence."}],
   }

   antwort = requests.post(url, json=payload, headers=headers, timeout=120)
   print("Status:", antwort.status_code)
   if antwort.ok:
       print("Antwort der KI:", antwort.json()["choices"][0]["message"]["content"])
   else:
       print("Fehlermeldung:", antwort.text[:500])
```
